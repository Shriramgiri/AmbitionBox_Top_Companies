```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
from curl_cffi import requests
import re
```

```python
url  = "https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav"
response = requests.get(url ,impersonate="chrome110")
soup = BeautifulSoup(response.text)
response
```

```python
name_data =[]
rating_data =[]
review_data =[]
salary_data =[]
job_data =[]
type_data=[]
highly_rated_data=[]
critically_rated_data=[]
for i in range (1,51):
    url =f"https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={i}"
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/123.0.0.0 Safari/537.36"}

    response = requests.get(url,headers=headers)
    html_text = response.text
    soup = BeautifulSoup(html_text)

    containers = soup.find_all("div", attrs ={"class":"companyCardWrapper"})
    
    for c  in containers :
        name = c.find("h2",attrs = {"class":"companyCardWrapper__companyName"})
        if name is None :
            name_data.append(np.nan)
        else:
            name_data.append(name.text)

        rating = c.find("div",attrs = {"class":"rating_text rating_text--md"})
        if rating is None :
            rating_data.append(np.nan)
        else:
            rating_data.append(rating.text)

        review = c.find("a", href=lambda x: x and "reviews" in x)
        if review:
            review_count = review.find("span", class_="companyCardWrapper__ActionCount")
            review_data.append(review_count.text.strip() if review_count else np.nan)
        else:
            review_data.append(np.nan)

        
        salary = c.find("a", href=lambda x: x and "salaries" in x)
        if salary:
            salary_count = salary.find("span", class_="companyCardWrapper__ActionCount")
            salary_data.append(salary_count.text.strip() if salary_count else np.nan)
        else:
            salary_data.append(np.nan)

        job= c.find("a", href=lambda x: x and "jobs" in x)
        if job:
            job_count = job.find("span", class_="companyCardWrapper__ActionCount")
            job_data.append(job_count.text.strip() if job_count else np.nan)
        else:
            salary_data.append(np.nan)

        c_type = c.find("span",attrs = {"class":"companyCardWrapper__interLinking"})
        if c_type is None :
            type_data.append(np.nan)
        else:
            type_data.append(c_type.text)

        ratings = c.find_all("span", class_="companyCardWrapper__ratingValues")

        highly = ratings[0].text.strip() if len(ratings) > 0 else np.nan
        critically = ratings[1].text.strip() if len(ratings) > 1 else np.nan

        highly_rated_data.append(highly)
        critically_rated_data.append(critically)
```

```python
print(len(name_data))
print(len(rating_data))
print(len(review_data))
print(len(salary_data))
print(len(highly_rated_data))
```

```python
df = pd.DataFrame({"name":name_data,
                   "rating":rating_data,
                   "review":review_data,
                   "salary":salary_data,
                   "job":job_data,
                   "type":type_data,
                   "highly_rated":highly_rated_data,
                   "critically_rated":critically_rated_data
                  })
```

```python
name_text = " "
for i in df["name"]:
    name_text = name_text + i + "/n"
```

```python
def extract_name(x):
    if pd.isna(x):
        return np.nan
    matches = re.findall(r"\s[A-Za-z]+", str(x))
    return matches[0].strip() if matches else str(x).strip()

df["name"] = df["name"].apply(extract_name)

# 2. Inspect ratings text
rating_text = ""
for i in df["rating"]:
    rating_text = rating_text + str(i) + "\n"
print(rating_text)

# Clean the 'rating' column safely
def extract_rating(x):
    if pd.isna(x):
        return np.nan
    matches = re.findall(r"[0-9]\.[0-9]", str(x))
    return matches[0] if matches else np.nan

df["rating"] = df["rating"].apply(extract_rating)

# 3. Inspect type text
type_text = ""
for i in df["type"]:
    print(str(i), "\n")

# Clean 'type' into 'type1' and 'location' safely
def extract_type1(x):
    if pd.isna(x):
        return np.nan
    matches = re.findall(r"^\s*(.*?)\s\|", str(x))
    return matches[0] if matches else np.nan

def extract_location(x):
    if pd.isna(x):
        return np.nan
    matches = re.findall(r"\|\s\w+", str(x))
    return matches[0] if matches else np.nan

df["type1"] = df["type"].apply(extract_type1)
df["location"] = df["type"].apply(extract_location)
df.drop(columns="type", inplace=True)


df
```

```python
df.rename(columns = {"type": "sector"}, inplace = True)
```

```python
# 1. Clean up whitespace/newlines from the sector column
df["sector"] = df["sector"].apply(lambda x: str(x).strip() if pd.notna(x) else np.nan)

# 2. Split into 'sector' and 'location' using the pipe separator (|)
def extract_sector(x):
    if pd.isna(x):
        return np.nan
    match = re.search(r"^(.*?)\s*\|", x)
    return match.group(1).strip() if match else x

def extract_location(x):
    if pd.isna(x):
        return np.nan
    match = re.search(r"\|\s*(.*)", x)
    return match.group(1).strip() if match else np.nan

df["clean_sector"] = df["sector"].apply(extract_sector)
df["location"] = df["sector"].apply(extract_location)

# Replace old sector column and clean up
df.drop(columns=["sector"], inplace=True)
df.rename(columns={"clean_sector": "sector"}, inplace=True)

df
```

```python

# Clean the 'location' column to keep only the primary city name
def extract_clean_location(x):
    if pd.isna(x):
        return np.nan
    # Match everything before the '+' sign or extra whitespace
    match = re.search(r"^(.*?)\s*\+", str(x))
    if match:
        return match.group(1).strip()
    # Fallback in case there is no '+' sign
    return str(x).strip()

df["location"] = df["location"].apply(extract_clean_location)

# Inspect the cleaned DataFrame
df.head()
```
